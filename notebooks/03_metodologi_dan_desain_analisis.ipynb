## 3. Konteks Singkat dan Landasan Ilmiah
Desain metodologi adalah cetak biru saintifik dari keseluruhan penelitian. Kesalahan dalam menetapkan baseline klimatologi atau pemilihan uji statistik yang melanggar asumsi data akan merusak validitas kesimpulan. Deteksi MHW memerlukan penentuan periode klimatologi dasar yang konsisten (misal 30 tahun) serta ambang batas persentil musiman (biasanya persentil ke-90 berbasis jendela 11 hari). Sementara itu, klorofil-a memiliki sifat distribusi yang asimetris sehingga penggunaan rata-rata aritmatika seringkali membiaskan interpretasi dibanding median. Untuk data perikanan, perbedaan resolusi tahunan menuntut pemilihan metodologi agregasi yang tepat agar tidak menimbulkan korelasi lancung (*spurious correlation*).

## 5. Instruksi Langkah demi Langkah
### Langkah 1: Definisi Operasional Variabel
Rumuskan definisi matematis dan satuan untuk setiap variabel: SST (°C), Ambang Batas MHW (°C), Intensitas MHW (°C dan °C-hari), Klorofil-a (mg/m³), Anomali Klorofil-a (mg/m³ atau %), Produksi Ikan (ton/tahun atau kg/bulan).
### Langkah 2: Formulasi Deteksi Kejadian MHW
Tuliskan formula acuan Hobday et al. (2016):
- Ambang batas: $T_{90}(j) = 	ext{Percentile}_{90}\left(\{SST(y, d) \mid d \in [j-5, j+5]\}ight)$
- Syarat MHW: $SST(t) > T_{90}(t)$ selama minimal 5 hari berturut-turut.
- Metrik: Onset date, Demise date, Durasi ($D$), Intensitas Maksimum ($I_{max}$), Intensitas Kumulatif ($I_{cum}$).
### Langkah 3: Desain Komparasi Klorofil-a
Rancang perbandingan klorofil-a pada jendela waktu MHW terhadap nilai baseline non-MHW pada bulan yang sama untuk mengontrol siklus musiman.
### Langkah 4: Pemilihan Uji Statistik
Tetapkan alur uji statistika:
1. Uji normalitas data klorofil-a dan data perikanan.
2. Jika tidak normal, gunakan statistik non-parametrik (Median, IQR, Mann-Whitney U test, Wilcoxon signed-rank).
3. Untuk analisis hubungan temporal, gunakan koefisien korelasi rank Spearman ($r_s$).
### Langkah 5: Penanganan Skala Waktu yang Berbeda
Rancang protokol: MHW dan Klorofil-a dianalisis pada skala event dan bulanan. Untuk menghubungkan dengan produksi ikan tahunan, hitung metrik tahunan agregat MHW (total hari MHW per tahun, intensitas kumulatif tahunan) dan rata-rata tahunan klorofil-a.
### Langkah 6: Daftar Batasan & Mitigasi Bias
Tuliskan batasan bahwa analisis korelasi tidak dapat membuktikan mekanisme sebab-akibat langsung, melainkan asosiasi kovariansi.
