## 3. Konteks Singkat dan Landasan Ilmiah
Keberhasilan analisis empiris sangat bergantung pada ketersediaan data nyata yang konsisten. Dalam penelitian kelautan multivariat di Indonesia, seringkali terjadi ketimpangan resolusi: data suhu permukaan laut (SST) tersedia harian dengan rekor panjang (>40 tahun), data klorofil-a satelit optik tersedia harian/8-harian namun terpotong masalah tutupan awan dan batasan masa hidup instrumen (misal MODIS Aqua sejak pertengahan 2002), sedangkan data statistik perikanan tangkap dari KKP atau BPS umumnya hanya tersedia dalam agregasi tahunan pada level kabupaten atau provinsi. Audit ketersediaan data wajib dilakukan secara ketat untuk menghindari perancangan metode analisis yang tidak dapat dieksekusi.


## 4. Daftar Tugas Kerja
Berikut adalah tugas konkret yang wajib diselesaikan pada notebook ini:
1. Mengidentifikasi kandidat produk data SST (misal: NOAA OISST v2.1 0.25°, ESA SST CCI, OSTIA).
2. Mengidentifikasi kandidat produk klorofil-a satelit (misal: NASA MODIS-Aqua L3, VIIRS, ESA OC-CCI).
3. Mengidentifikasi sumber resmi data produksi cakalang dan tongkol di NTT (KKP Satu Data, BPS NTT Dalam Angka, DKP NTT).
4. Mendokumentasikan metadata teknis: rentang tahun, resolusi temporal, resolusi spasial, format berkas, satuan, lisensi penggunaan, dan URL akses resmi.
5. Memeriksa konsistensi pelaporan data perikanan (pemisahan cakalang vs tongkol, konsistensi nama kabupaten pemekaran, kontinuitas tahun).
6. Mencatat isu nilai kosong (*missing values*), diskontinuitas sensor satelit, dan perubahan metodologi sensus/pencatatan data ikan.
7. Menganalisis irisan rentang waktu (*temporal overlap*) antar-variabel dan merekomendasikan periode penelitian optimal.
8. Menetapkan status verifikasi tiap dataset (Belum diperiksa / Ditemukan / Akses terverifikasi / Tidak memadai).


## 5. Instruksi Langkah demi Langkah
### Langkah 1: Audit Produk Suhu Muka Laut (SST)
Evaluasi produk NOAA High-Resolution Daily OISST v2.1. Cek: rentang waktu (1981-sekarang), grid (0.25 x 0.25 derajat), format NetCDF, satuan Celcius. Tinjau apakah resolusi 0.25 derajat memadai untuk selat di NTT.
### Langkah 2: Audit Produk Satelit Klorofil-a
Evaluasi produk NASA MODIS-Aqua L3 (4 km, 2002-sekarang) dan ESA OC-CCI (multi-sensor merged, 1997-sekarang). Periksa prevalensi missing data akibat tutupan awan monsun di NTT.
### Langkah 3: Audit Data Statistik Perikanan
Telusuri portal Satu Data KKP dan katalog BPS Provinsi NTT. Periksa ketersediaan data produksi (ton) untuk cakalang (*Katsuwonus pelamis*) dan tongkol (*Euthynnus affinis* / *Auxis thazard*). Catat apakah data tersedia bulanan di pelabuhan pendaratan atau hanya tahunan di tingkat kabupaten.
### Langkah 4: Pengisian Inventaris Data
Isi tabel matriks inventarisasi dengan template 11 kolom wajib: Variabel | Produk/sumber | Periode | Resolusi waktu | Resolusi spasial | Cakupan | Format | Satuan | URL | Status verifikasi | Catatan.
### Langkah 5: Penentuan Irisan Periode Waktu
Buat bagan atau garis waktu irisan data untuk melihat periode tumpang tindih terpanjang di mana ketiga dataset tersedia secara simultan.
### Langkah 6: Perumusan Rekomendasi Periode
Rekomendasikan rentang tahun penelitian definitif berbasis bukti audit (misal: 2003–2022 jika mengikuti MODIS Aqua dan data BPS konsisten).
