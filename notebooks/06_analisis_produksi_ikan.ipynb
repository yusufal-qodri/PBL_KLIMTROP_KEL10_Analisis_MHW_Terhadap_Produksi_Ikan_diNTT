## 3. Konteks Singkat dan Landasan Ilmiah
Perikanan tangkap cakalang dan tongkol merupakan komoditas unggulan nelayan di Nusa Tenggara Timur, khususnya di perairan Laut Sawu, Laut Flores, dan Laut Timor. Namun, dalam analisis dinamika perikanan menggunakan data statistik resmi (KKP dan BPS), peneliti dihadapkan pada tantangan pelaporan: definisi data seringkali bercampur antara volume produksi riil perairan, pendaratan di pelabuhan pendaratan ikan (PPI), dan hasil tangkapan yang didaratkan di luar wilayah tangkap. Selain itu, dinamika produksi ikan dipengaruhi oleh faktor-faktor non-oseanografi yang sangat kuat, seperti upaya penangkapan (*fishing effort*), jumlah armada kapal motor, ketersediaan BBM solar subsidi, cuaca buruk gelombang tinggi, serta dinamika harga pasar. Analisis produksi ikan harus dirancang sebagai studi tren yang teliti tanpa melompat pada kesimpulan bahwa fluktuasi produksi semata-mata dikendalikan oleh anomali iklim laut.

## 5. Instruksi Langkah demi Langkah
### Langkah 1: Standardisasi Definisi dan Nomenklatur
Tetapkan definisi istilah sesuai KKP: 'Produksi Perikanan Tangkap' adalah volume ikan yang ditangkap dan didaratkan untuk tujuan komersial maupun konsumsi. Pisahkan pencatatan cakalang (*skipjack tuna*) dari tongkol (*eastern little tuna / frigate tuna*).
### Langkah 2: Penelaahan Format Data Tabulasi
Rancang tabel standardisasi data statistik: `tahun`, `kode_wilayah`, `nama_wilayah`, `nama_spesies`, `nama_ilmiah`, `volume_produksi_ton`, `sumber_publikasi`, `status_verifikasi`.
### Langkah 3: Pemeriksaan Kualitas dan Diskontinuitas
Susun checklist pemeriksaan diskontinuitas data: periksa apakah ada lonjakan tajam >200% dalam satu tahun yang disebabkan oleh penambahan cakupan pelabuhan baru atau perubahan metodologi survei BPS.
### Langkah 4: Desain Analisis Tren Temporal
Rancang prosedur analisis deret waktu: visualisasi grafik garis tahunan (*time series plot*), perhitungan laju pertumbuhan rata-rata tahunan (*compound annual growth rate*), dan pemisahan tren cakalang vs tongkol.
### Langkah 5: Dokumentasi Faktor Perancu Antropogenik
Susun tabel faktor perancu perikanan tangkap NTT: upaya penangkapan (jumlah kapal, trip melaut), harga pasar, regulasi zona tangkap, cuaca monsun, dan logistik pendaratan.

Tuliskan klausul batasan ilmiah: fluktuasi produksi tidak dapat diklaim sebagai akibat langsung dari variabilitas MHW atau klorofil-a tanpa data upaya penangkapan (*effort-standardized CPUE*).
