## 3. Konteks Singkat dan Landasan Ilmiah
Klorofil-a merupakan pigmen fotosintetik utama pada fitoplankton dan berfungsi sebagai indikator kelimpahan biomassa produsen primer di lapisan permukaan laut. Dalam penginderaan jauh laut, konsentrasi klorofil-a diestimasi melalui rasio pantulan spektral cahaya biru-hijau. Di perairan tropis NTT, data satelit optik menghadapi tantangan signifikan berupa tutupan awan yang tebal saat Monsun Barat, yang menyebabkan tingginya persentase piksel kosong (*missing pixels*). Selain itu, klorofil-a bukan merupakan pengukuran langsung terhadap nutrien terlarut (seperti nitrat atau fosfat) maupun biomassa ikan. Analisis klorofil-a harus dirancang dengan kehati-hatian statistik yang tinggi.
