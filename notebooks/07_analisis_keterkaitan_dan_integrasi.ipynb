## 5. Instruksi Langkah demi Langkah
### Langkah 1: Evaluasi Kompatibilitas Skala (Scale Matching)
Analisis apakah data dapat diagregasi secara sah. Tetapkan aturan: MHW (harian) dan Klorofil-a (harian/8-harian) kompatibel untuk analisis skala kejadian (*event-scale*). Data perikanan tahunan hanya dapat disandingkan pada skala metrik tahunan agregat (misal: Total Durasi MHW Tahunan, Anomali Klorofil-a Rerata Tahunan, Produksi Ikan Tahunan).
### Langkah 2: Perumusan Skema Tabel Integrasi Multivariat
Rancang tabel ringkasan tahunan: `tahun`, `frekuensi_event_mhw`, `total_hari_mhw`, `intensitas_kumulatif_mhw_degC_days`, `anomali_median_chla`, `produksi_cakalang_ton`, `produksi_tongkol_ton`, `catatan_kualitas_data`.
### Langkah 3: Prosedur Pemilihan Uji Statistik Asosiasi
Tentukan kriteria uji: Jika jumlah tahun pengamatan $N < 30$, uji normalitas seringkali memiliki daya rendah; gunakan koefisien korelasi non-parametrik Spearman rank ($r_s$) dan sertakan interval kepercayaan (*confidence interval*).
### Langkah 4: Penegasan Larangan Spekulasi Kausalitas
Tuliskan aturan ilmiah baku: Korelasi signifikan antara anomali termal dan penurunan produksi ikan TIDAK membuktikan bahwa MHW membunuh atau mengusir ikan, melainkan hanya menunjukkan kovariansi statistik yang harus ditelaah bersama faktor upaya tangkap dan migrasi.
### Langkah 5: Protokol Kontinjensi Data Tidak Memadai
Tetapkan aturan: Jika data produksi ikan memiliki terlalu banyak missing data, diskontinuitas ekstrem, atau N < 10 tahun, batalkan uji korelasi kuantitatif trilateral. Alihkan fokus analisis produksi ikan menjadi telaah kualitatif-deskriptif yang disandingkan dengan temuan kuantitatif MHW-Klorofil-a.
### Langkah 6: Penyusunan Format Sintesis Diskusi
Rancang alur pembahasan: (1) Temuan respon klorofil-a terhadap MHW; (2) Tren produksi perikanan; (3) Evaluasi kemungkinan keterkaitan biologis-oseanografi; (4) Daftar faktor pembatas interpretasi.
