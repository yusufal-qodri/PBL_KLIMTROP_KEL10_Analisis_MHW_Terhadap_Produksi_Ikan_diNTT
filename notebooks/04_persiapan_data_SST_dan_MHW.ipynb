## 3. Konteks Singkat dan Landasan Ilmiah
Data SST harian gridded (seperti NOAA OISST) merupakan input primer untuk mendeteksi gelombang panas laut. Proses persiapan data memerlukan manipulasi array multidimensi (waktu, garis lintang, garis bujur). Sebelum algoritma dijalankan, struktur berkas NetCDF harus diperiksa integritasnya: koordinat spasial, unit suhu (°C), sistem kalender gregorian, serta deteksi nilai kosong (NaN / flag daratan). Kejadian MHW kemudian diidentifikasi per piksel atau per rata-rata spasial area NTT, menghasilkan tabel ringkasan kejadian MHW yang terstruktur.


## 4. Daftar Tugas Kerja
Berikut adalah tugas konkret yang wajib diselesaikan pada notebook ini:
1. Merancang alur inspeksi struktur berkas data SST (dimensi waktu, latitude, longitude, missing values, fill value daratan).
2. Merancang prosedur pemotongan spasial (*spatial cropping*) sesuai bounding box perairan NTT.
3. Merancang alur perhitungan rata-rata klimatologi harian dan ambang batas persentil ke-90 menggunakan jendela berjalan 11-hari.

6. Menyusun protokol kendali mutu (*quality assurance*) untuk memastikan tidak ada event palsu akibat artifact data satelit.
7. Mendokumentasikan asumsi teknis dan potensi kendala perhitungan numerik.


## 5. Instruksi Langkah demi Langkah
### Langkah 1: Rancangan Struktur Input SST
Tetapkan variabel input: `time`, `lat`, `lon`, `sst`. Rancang pemeriksaan integritas data: rentang nilai SST realistis untuk perairan tropis NTT (antara 24°C hingga 33°C), identifikasi kode missing value.
### Langkah 2: Rancangan Pemotongan Spasial
Tentukan koordinat pemotongan: latitude [-12.5° s.d. -8.0°], longitude [118.0° s.d. 126.0°]. Rencanakan pemisahan antara piksel daratan pulau (Flores, Sumba, Timor) dan piksel laut.
### Langkah 3: Rancangan Perhitungan Klimatologi Harian
Rancang algoritma pembentukan nilai klimatologi harian $T_{clim}(d)$ dan ambang batas $T_{90}(d)$ untuk setiap hari dalam kalender (1–365) menggunakan jendela 11 hari selama periode dasar klimatologi.
### Langkah 4: Rancangan Algoritma Identifikasi Event
Rancang logika penentuan event MHW: deteksi segmen di mana anomali terhadap ambang batas bernilai positif selama minimal 5 hari berturut-turut. Hubungkan event yang berjarak <= 2 hari.
### Langkah 5: Perancangan Skema Kolom Tabel Katalog Event
Rancang struktur tabel event MHW: `event_id`, `start_date`, `end_date`, `duration_days`, `max_intensity_degC`, `mean_intensity_degC`, `cum_intensity_degC_days`, `category` (Moderate, Strong, Severe, Extreme).
### Langkah 6: Penyusunan Checklist Validasi Teknis
Susun checklist untuk memverifikasi apakah event terdeteksi konsisten secara fisik dan tidak terdistorsi oleh satu piksel anomali.
