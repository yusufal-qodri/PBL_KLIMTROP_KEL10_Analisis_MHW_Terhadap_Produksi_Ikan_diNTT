## 3. Konteks Singkat dan Landasan Ilmiah
Wilayah perairan Nusa Tenggara Timur (NTT) merupakan zona oseanografi strategis di Indonesia bagian timur yang dilewati sistem Arus Lintas Indonesia (ARLINDO) dan kaya akan keanekaragaman hayati laut. Perairan ini menopang perikanan pelagis penting, khususnya cakalang (*Katsuwonus pelamis*) dan tongkol (*Euthynnus affinis* / *Auxis spp.*). Namun, meningkatnya frekuensi dan intensitas cuaca laut ekstrem seperti Marine Heatwaves (MHW) akibat perubahan iklim global berpotensi memberikan tekanan termal terhadap ekosistem perairan, memodifikasi dinamika stratifikasi massa air, menekan konsentrasi klorofil-a (fitoplankton), dan berpotensi memengaruhi distribusi serta hasil tangkapan perikanan.
