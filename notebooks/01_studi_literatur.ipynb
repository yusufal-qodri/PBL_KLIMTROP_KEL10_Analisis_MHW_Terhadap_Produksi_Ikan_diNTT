## 3. Konteks Singkat dan Landasan Ilmiah
Studi literatur yang komprehensif adalah pondasi dari metodologi yang valid. Definisi operasional MHW paling banyak dirujuk secara global mengikuti kerangka kerja Hobday et al. (2016), namun penerapannya di perairan tropis Indonesia memerlukan penyesuaian terkait baseline klimatologi, variabilitas monsunal, fenomena ENSO (*El Niño-Southern Oscillation*), dan IOD (*Indian Ocean Dipole*). Selain itu, hubungan antara anomali suhu dan klorofil-a tidak selalu linear; pemanasan permukaan dapat memicu penipisan lapisan tercampur (*mixed layer*) dan penurunan upwelling nutrien, yang menekan pertumbuhan fitoplankton. Pada sisi perikanan, cakalang dan tongkol adalah spesies pelagis yang sangat mobile, sehingga fluktuasi produksi di pelabuhan pendaratan NTT harus ditinjau berdasarkan literatur perikanan tangkap regional.
